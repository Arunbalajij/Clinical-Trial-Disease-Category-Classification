```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# -------------------------------------------------------------
# 1. Load Dataset
# -------------------------------------------------------------
df = pd.read_csv('clinical_trials_raw_patient2trial_conditions.csv')
df.columns = df.columns.str.strip().str.lower()

# -------------------------------------------------------------
# 2. Handle Missing Values and Duplicate Records
# -------------------------------------------------------------
# Check initial counts
print(f'Initial total rows: {len(df)}')

# Remove duplicate rows (if any) based on text and condition
df = df.drop_duplicates(subset=['brief_summary', 'source_condition_query'])

# Drop rows where essential columns are NaN
df = df.dropna(subset=['brief_summary', 'source_condition_query']).copy()

# Reset index after dropping rows
df.reset_index(drop=True, inplace=True)
print(f'Cleaned row count after dropping nulls & duplicates: {len(df)}')

# -------------------------------------------------------------
# 3. Text Cleaning, Normalization, Tokenization & Lemmatization
# -------------------------------------------------------------
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))


def preprocess_text(text):
  # Normalization: Lowercase conversion
  text = str(text).lower()

  # Remove special characters, digits, and punctuation (keep only letters)
  text = re.sub(r'[^a-z\s]', ' ', text)

  # Tokenization
  tokens = word_tokenize(text)

  # Stop word removal and Lemmatization
  cleaned_tokens = [
      lemmatizer.lemmatize(word)
      for word in tokens
      if word not in stop_words and len(word) > 2
  ]

  # Rejoin tokens into a single cleaned string for vectorization
  return ' '.join(cleaned_tokens)


# Apply the pipeline to the text column
df['processed_summary'] = df['brief_summary'].apply(preprocess_text)

# Inspect sample result
print('\nSample Raw Summary:')
print(df['brief_summary'].iloc[0][:150], '...')
print('\nSample Processed Summary:')
print(df['processed_summary'].iloc[0][:150], '...')

# -------------------------------------------------------------
# 4. Encode Target Labels for Machine Learning Models
# -------------------------------------------------------------
label_encoder = LabelEncoder()
df['label_encoded'] = label_encoder.fit_transform(df['source_condition_query'])

# Display mapping between class names and assigned integers
class_mapping = dict(
    zip(label_encoder.classes_, range(len(label_encoder.classes_)))
)
print('\nTarget Class Mapping:')
for label, idx in class_mapping.items():
  print(f'  {idx}: {label}')

# -------------------------------------------------------------
# 5. Split Dataset (Train / Test)
# -------------------------------------------------------------
# Stratification ensures proportional class representation in both sets
X = df['processed_summary']
y = df['label_encoded']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# -------------------------------------------------------------
# 6. TF-IDF Vectorization
# -------------------------------------------------------------
# Fit vectorizer only on train data to prevent data leakage; transform test
tfidf_vectorizer = TfidfVectorizer(
    max_features=5000,  # Limits vocabulary size to top informative terms
    ngram_range=(1, 2),  # Captures unigrams and bigrams (e.g., 'breast cancer')
    min_df=2,  # Ignores terms appearing in fewer than 2 documents
    max_df=0.95,  # Ignores terms appearing in more than 95% of documents
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print(f'\nTF-IDF Train Matrix Shape: {X_train_tfidf.shape}')
print(f'TF-IDF Test Matrix Shape:  {X_test_tfidf.shape}')
```

```python
from collections import Counter
import re
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

# Define clinical stop words to remove generic trial vocabulary
clinical_stopwords = set(stopwords.words('english')).union({
    'study',
    'trial',
    'patient',
    'patients',
    'treatment',
    'clinical',
    'purpose',
    'investigate',
    'determine',
    'evaluate',
    'efficacy',
    'safety',
    'group',
    'arm',
    'phase',
})


def clean_for_eda(text):
  text = re.sub(r'[^a-zA-Z\s]', ' ', str(text).lower())
  return ' '.join(
      [w for w in text.split() if w not in clinical_stopwords and len(w) > 2]
  )


df['eda_summary'] = df['brief_summary'].apply(clean_for_eda)


# Top N-Grams extraction helper
def get_top_ngrams(corpus, ngram_range=(1, 1), top_n=15):
  vec = CountVectorizer(ngram_range=ngram_range).fit(corpus)
  bag_of_words = vec.transform(corpus)
  sum_words = bag_of_words.sum(axis=0)
  words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
  words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
  return pd.DataFrame(words_freq[:top_n], columns=['Keyword / N-Gram', 'Count'])


# Top Unigrams and Bigrams
top_unigrams = get_top_ngrams(df['eda_summary'], ngram_range=(1, 1), top_n=10)
top_bigrams = get_top_ngrams(df['eda_summary'], ngram_range=(2, 2), top_n=10)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(
    data=top_unigrams,
    x='Count',
    y='Keyword / N-Gram',
    ax=axes[0],
    palette='Blues_r',
)
axes[0].set_title('Top 10 Clinical Keywords (Unigrams)')

sns.barplot(
    data=top_bigrams,
    x='Count',
    y='Keyword / N-Gram',
    ax=axes[1],
    palette='Oranges_r',
)
axes[1].set_title('Top 10 Clinical Key Phrases (Bigrams)')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import LinearSVC

# -------------------------------------------------------------
# 1. Prepare Data and Split
# -------------------------------------------------------------
# Ensure missing values are removed and indices reset
df = df.dropna(
    subset=['brief_summary', 'source_condition_query']
).drop_duplicates(subset=['brief_summary', 'source_condition_query'])
df = df.reset_index(drop=True)

# Encode disease category labels
label_encoder = LabelEncoder()
df['target'] = label_encoder.fit_transform(df['source_condition_query'])
class_names = list(label_encoder.classes_)

# 80/20 Stratified train-test split
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    df['brief_summary'],
    df['target'],
    test_size=0.20,
    random_state=42,
    stratify=df['target'],
)

# -------------------------------------------------------------
# 2. TF-IDF Feature Extraction
# -------------------------------------------------------------
# Fit ONLY on the training split to avoid data leakage
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    stop_words='english',
    sublinear_tf=True,  # Applies log scaling to term frequencies
)

X_train = tfidf.fit_transform(X_train_raw)
X_test = tfidf.transform(X_test_raw)

print(f'Train feature matrix shape: {X_train.shape}')
print(f'Test feature matrix shape:  {X_test.shape}')

# -------------------------------------------------------------
# 3. Define Classifiers to Compare
# -------------------------------------------------------------
models = {
    'Multinomial Naive Bayes': MultinomialNB(alpha=0.1),
    'Logistic Regression': LogisticRegression(
        max_iter=1000, class_weight='balanced', random_state=42
    ),
    'Linear SVM': LinearSVC(
        class_weight='balanced', random_state=42, dual='auto'
    ),
    'Random Forest': RandomForestClassifier(
        n_estimators=150, max_depth=30, random_state=42, n_jobs=-1
    ),
}

# -------------------------------------------------------------
# 4. Train, Predict, and Evaluate
# -------------------------------------------------------------
results = []
predictions = {}

for name, model in models.items():
  print(f'Training: {name}...')
  model.fit(X_train, y_train)
  y_pred = model.predict(X_test)
  predictions[name] = y_pred

  # Aggregate classification metrics
  acc = accuracy_score(y_test, y_pred)
  prec = precision_score(y_test, y_pred, average='weighted', zero_division=0)
  rec = recall_score(y_test, y_pred, average='weighted', zero_division=0)
  f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

  results.append({
      'Model': name,
      'Accuracy': acc,
      'Precision (Weighted)': prec,
      'Recall (Weighted)': rec,
      'F1-Score (Weighted)': f1,
  })

# Display summary performance table
df_results = (
    pd.DataFrame(results)
    .sort_values(by='F1-Score (Weighted)', ascending=False)
    .reset_index(drop=True)
)
print('\n--- Model Comparison Table ---')
print(df_results.to_string(index=False))
```

```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
)
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder

# -------------------------------------------------------------------------
# Step 1: Data Preparation & Target Encoding
# -------------------------------------------------------------------------
# Load clean dataset
df = pd.read_csv("clinical_trials_raw_patient2trial_conditions.csv")
df.columns = df.columns.str.strip().str.lower()

# Clean missing and duplicate values
df = df.dropna(subset=["brief_summary", "source_condition_query"])
df = df.drop_duplicates(subset=["brief_summary", "source_condition_query"]).reset_index(drop=True)

X = df["brief_summary"]
y_raw = df["source_condition_query"]

# Encode target categorical classes to integers
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y_raw)
class_names = [str(cls) for cls in label_encoder.classes_]

# 80/20 Stratified train-test split to preserve class ratios
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# -------------------------------------------------------------------------
# Step 2: Build the NLP Classification Pipeline
# -------------------------------------------------------------------------
model_pipeline = Pipeline([
    (
        "tfidf",
        TfidfVectorizer(
            max_features=25000,
            ngram_range=(1, 3),      # Unigrams, bigrams, and trigrams
            sublinear_tf=True,       # 1 + log(tf) scaling to suppress repetitive medical jargon
            min_df=2,                # Drop rare typos / single occurrences
            max_df=0.90,             # Drop terms appearing in >90% of documents
            stop_words="english",
        ),
    ),
    (
        "classifier",
        LogisticRegression(
            C=2.0,                   # Tuned regularization strength
            max_iter=1500,
            class_weight="balanced", # Handles potential disease class imbalances
            random_state=42,
            solver="lbfgs",
        ),
    ),
])

# -------------------------------------------------------------------------
# Step 3: Cross-Validation & Model Training
# -------------------------------------------------------------------------
print("Running 5-Fold Stratified Cross-Validation...")
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(model_pipeline, X_train, y_train, cv=cv, scoring="accuracy", n_jobs=-1)
print(f"5-Fold CV Accuracy: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")

# Fit on full training split
model_pipeline.fit(X_train, y_train)
```

```python
# Predictions on holdout test set
y_pred = model_pipeline.predict(X_test)

test_acc = accuracy_score(y_test, y_pred)
test_f1 = f1_score(y_test, y_pred, average="weighted")

print(f"\nHoldout Test Accuracy: {test_acc:.4f}")
print(f"Holdout Test F1-Score (Weighted): {test_f1:.4f}\n")

# Detailed class-level precision, recall, and F1
print("Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=class_names, digits=4))

# Normalized Confusion Matrix Heatmap
cm = confusion_matrix(y_test, y_pred, normalize="true")
plt.figure(figsize=(max(8, len(class_names) * 0.9), max(6, len(class_names) * 0.7)))
sns.heatmap(
    cm,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names,
)
plt.title("Normalized Confusion Matrix - Disease Category Classification")
plt.xlabel("Predicted Disease Label")
plt.ylabel("True Disease Label")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
def classify_clinical_summary(summary_text: str, pipeline, encoder, top_k: int = 3):
    """
    Predicts the top disease categories with confidence probabilities.
    """
    # Predict class probabilities
    probabilities = pipeline.predict_proba([summary_text])[0]
    
    # Sort top K predictions
    top_indices = np.argsort(probabilities)[::-1][:top_k]
    
    results = []
    for idx in top_indices:
        results.append({
            "disease_category": encoder.inverse_transform([idx])[0],
            "confidence": float(probabilities[idx])
        })
    return results

# Sample test run
sample_text = (
    "A randomized, double-blind trial to evaluate insulin sensitivity, fasting "
    "plasma glucose, and HbA1c levels in adult patients receiving metformin "
    "combined with a novel SGLT2 inhibitor."
)

predictions = classify_clinical_summary(sample_text, model_pipeline, label_encoder)

print("--- Inference Result ---")
for rank, pred in enumerate(predictions, 1):
    print(f"{rank}. {pred['disease_category']} - Confidence: {pred['confidence']:.2%}")
```
